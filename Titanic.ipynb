```python
import pandas as pd 
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv" 
df = pd.read_csv(url) 
df.head()
```

```python
df.info() 
df.describe() 
df.isnull().sum()
```

```python
df["Survived"].mean()                       # overall survival rate 
df.groupby("Sex")["Survived"].mean()        # survival by sex 
df.groupby("Pclass")["Survived"].mean()     # survival by class 
```
